```python
from selenium import webdriver
```

```python
driver = webdriver.Chrome()
```

```python
driver.get("https://www.imdb.com/chart/top/")
```

```python
movies = driver.find_elements(
    "css selector",
    "li.ipc-metadata-list-summary-item"
)

print(len(movies))
```

```python
ranks = []
movie_names = []

for movie in movies:
    rank = movie.find_element("css selector", "div.ipc-signpost__text").text
    name = movie.find_element("css selector", "h4.ipc-title__text").text

    ranks.append(rank.replace("#", ""))
    movie_names.append(name)

print(ranks[:5])
print(movie_names[:5])
```
